# Python Insights - Analisando Dados com Python

### Case - Cancelamento de Clientes

Você foi contratado por uma empresa com mais de 800 mil clientes para um projeto de Dados. Recentemente a empresa percebeu que da sua base total de clientes, a maioria são clientes inativos, ou seja, que já cancelaram o serviço.

Precisando melhorar seus resultados ela quer conseguir entender os principais motivos desses cancelamentos e quais as ações mais eficientes para reduzir esse número.

A base de dados se encontra nesse repositório com o nome 'cancelamentos.csv'
É preciso instalar o Jupyter no VSCode pra executar.

```python
#What we do?
#import DB
#Visualize DB
#Fix DB issues
#Analysis of cancellations
#Analysis of the causes of cancellations (How do the columns cause impact cancellation?)

#import DB
import pandas as pd

cancellations_table = pd.read_csv("cancellations.csv")
display(cancellations_table)
```

```python
#Well... We don't need the 'CustomerID' column, so let's drop it
# cancellations_table.drop(columns=['CustomerID'], inplace=True) # It works too, but let's do it in a different way
cancellations_table = cancellations_table.drop("CustomerID", axis=1)
display(cancellations_table)
display(cancellations_table.info())
```

```python
#As we can see above, in the non-null column we have different values, meaning there are comlumns with empty values. 
#Let's drop the rows with empty values, as we don't want to analyze incomplete data.

#Looking more closely, we can see that the 'sexo' column has 3 empty values. Let's check what are these rows.
display(cancellations_table[cancellations_table["sexo"].isna()])

#And the 'tempo_como_cliente' column
display(cancellations_table[cancellations_table["tempo_como_cliente"].isna()])

#It's just to know more. So we have 4 rows (5, 12, 17 and 18) with empty values. Let's drop them.
cancellations_table = cancellations_table.dropna()

#Let's check the info of the table after dropping the rows with empty values.
display(cancellations_table.info())
```

```python
#The data base has already been cleaned, so we can start the analysis of the cancellations. 
# Let's check the number of cancellations and the number of non-cancellations.
display(cancellations_table["cancelou"].value_counts())

#And in terms of percentage?
print("Amount of cancellations and non-cancellations in percentage:")
display(cancellations_table["cancelou"].value_counts(normalize=True).map("{:.1%}".format)) 
```

```python
#What types of contract duration do we have?
display(cancellations_table["duracao_contrato"].unique())

#So we have 3 types of contract duration. Let's check the amount and percentage of cancellations for each type of contract.
display(cancellations_table.groupby("duracao_contrato")["cancelou"].value_counts().unstack())
display(cancellations_table.groupby("duracao_contrato")["cancelou"].value_counts(normalize=True).unstack().map("{:.1%}".format))

#Another way to view the data average, generally sorted by 'duracao_contrato' column.
display(cancellations_table.groupby("duracao_contrato").mean(numeric_only=True))
```

```python
#Let's see the graph representation of the number of CANCELLATIONS for each CONTRACT DURATION.
import plotly.express as px

graph_cancellations_by_contract = px.bar(cancellations_table.groupby("duracao_contrato")["cancelou"].value_counts().unstack(),
                                         text_auto=True, title="Number of Cancellations by Contract Duration" , labels={"value": "Number of contracts", "duracao_contrato": "Contract Duration", "cancelou": "Cancellation Status"})
display(cancellations_table)
graph_cancellations_by_contract.show()
```

```python
#Based on what we've seen so far, we can see that the 'duracao_contrato' column has a strong impact on the cancellation. 
#Montly contracts have a cancellation rate of 100%, while the other average 45%. 
#Therefore, we need to exclude the monthly contract from our analysis to clearly visualize the impact of the total cancellations volume. 
#Let's do that!

filtered_table_different_by_monthly = cancellations_table[cancellations_table["duracao_contrato"] != "Monthly"]
display(filtered_table_different_by_monthly["cancelou"].value_counts())

display(filtered_table_different_by_monthly.groupby("duracao_contrato")["cancelou"].value_counts(normalize=True).unstack().map("{:.1%}".format))

print("Amount of cancellations and non-cancellations in percentage:")
display(filtered_table_different_by_monthly["cancelou"].value_counts(normalize=True).map("{:.1%}".format)) 
```

```python
#Let's show the comparison of the cancellation rates BEFORE and AFTER filtering out the monthly contracts.

#Taking the cancellation rates before filtering out the monthly contracts.
resume_cancellations_before_monthly_filter = (
    cancellations_table["cancelou"]
    .value_counts(normalize=True)
    .rename("Before removing Monthly contracts")
)

#Taking the cancellation rates after filtering out the monthly contracts.
resume_cancellations_after_monthly_filter = (
    filtered_table_different_by_monthly["cancelou"]
    .value_counts(normalize=True)
    .rename("After removing Monthly contracts")
)

#Let's create a new table to compare the cancellation rates before and after filtering out the monthly contracts.
relation_cancellations = pd.concat(
    [resume_cancellations_before_monthly_filter, resume_cancellations_after_monthly_filter],
    axis=1
    ).reset_index()

#Let's rename the columns to make it more clear.
relation_cancellations = relation_cancellations.melt(
    id_vars="cancelou",
    var_name="Table",
    value_name="Percent"
)

#Let's create a new column to make it more clear.
relation_cancellations["status"] = relation_cancellations["cancelou"].map({
    0.0: "Active Clients",
    1.0: "Cancelled Clients"
})
display(relation_cancellations)

fig = px.bar(
    relation_cancellations,
    x="Table",
    y="Percent",
    color="status",
    barmode="stack",
    text_auto=".1%",
    title="Cancellation Rate — Before vs. After",
    color_discrete_map={
        "Cancelled Clients": "#E74C3C",
        "Active Clients": "#3498DB"
    },
    labels={
        "Percent": ""
    }
)

fig.show()

#For Managers: Ok, now that we've seen a 10.7% reduction in cancellations (56.8% -> 46.1%),
# we should consider offering a larger discount on the other plans to discourage users from choosing the monthly plan.
```

```python
#Let's get the result of table counting by subscription type
display(filtered_table_different_by_monthly["assinatura"].value_counts(normalize=True))
display(filtered_table_different_by_monthly.groupby("assinatura").mean(numeric_only=True))
#As we can see, the 'assinatura' hasen't a strong impact on the cancellation rate. 
#Let's do another graph to see if we can find a column that has a strong impact on the cancellation rate.

for column in filtered_table_different_by_monthly.columns:
    graph = px.histogram(filtered_table_different_by_monthly, x=column, color="cancelou", width=500)
    # graph.show()
```

```python
#So... We observe that cancellation rates are higher among people over the age of 50;
#Women have a higher cancellation rate than men;
#More than four calls to the call center results in cancellation;
#A payment delay of more than 20 days usually results in cancellation.

#Let's start with CALL CENTER calls, removing registers when call_center calls bigger than 5
filtered_table_different_by_monthly_and_calls = filtered_table_different_by_monthly[
    filtered_table_different_by_monthly["ligacoes_callcenter"]<5
    ]

#We can see the summary of cancellation for each call number
display(filtered_table_different_by_monthly_and_calls
        .groupby("ligacoes_callcenter")["cancelou"]
        .value_counts(normalize=True)
        .unstack()
        .map("{:.1%}".format))

#Showing two plain text rates, before and after call center filtering
display(filtered_table_different_by_monthly["cancelou"]
        .value_counts(normalize=True)
        .map("{:.1%}".format))
display(filtered_table_different_by_monthly_and_calls["cancelou"]
        .value_counts(normalize=True)
        .map("{:.1%}".format))

# #Let's create a new column to make it more clear.
# filtered_table_different_by_monthly_and_calls["status"] = filtered_table_different_by_monthly_and_calls["cancelou"].map({
#     0.0: "Active Clients",
#     1.0: "Cancelled Clients"
# })

graph_without_filtering_calls = px.histogram(filtered_table_different_by_monthly, x=column, color="cancelou", width=500,
                      title="Cancelled rate before improving the call center")
graph_without_filtering_calls.show()

graph_with_filtering_calls = px.histogram(filtered_table_different_by_monthly_and_calls, x=column, color="cancelou", width=500,
                     title="Cancelled rate after improving the call center")
graph_with_filtering_calls.show()

#For Managers: Ok, now that we've seen further 19.7% reduction in cancellations (46.1% -> 21.4),
#we should consider to improve the call center to offer definitive solutions.


```

```python
#A PAYMENT DELAY of more than 20 days usually results in cancellation.

filtered_table_different_by_monthly_and_calls_and_less_payment_delay = filtered_table_different_by_monthly_and_calls[
    filtered_table_different_by_monthly_and_calls["dias_atraso"] < 20
    ]

#Showing two plain text rates, before and after payment delay filtering
display(filtered_table_different_by_monthly_and_calls["cancelou"]
        .value_counts(normalize=True)
        .map("{:.1%}".format))
display(filtered_table_different_by_monthly_and_calls_and_less_payment_delay["cancelou"]
        .value_counts(normalize=True)
        .map("{:.1%}".format))
#Now we have 18.3% of cancellation rate.

graph_with_delay_filtering = px.histogram(filtered_table_different_by_monthly_and_calls_and_less_payment_delay, 
                                          x=column, 
                                          color="cancelou",
                                          width=500)
graph.show()

#For Managers: Maybe could we offer a discount for ponctual payments or even for early.
```

```python
#Let's see the final chart/ graph.

resume_cancellations = (
    cancellations_table["cancelou"]
    .value_counts(normalize=True)
    .rename("Original table")
)

#Taking the final cancellation rates.
resume_cancellations_with_monthly_filter = (
    filtered_table_different_by_monthly["cancelou"]
    .value_counts(normalize=True)
    .rename("After removing Monthly contracts")
)

resume_cancellations_with_monthly_filter_and_calls_filter = (
    filtered_table_different_by_monthly_and_calls["cancelou"]
    .value_counts(normalize=True)
    .rename("After removing more than 4 call center calls")
)

resume_cancellations_with_monthly_filter_and_calls_filter_and_payment_dalay_filter = (
    filtered_table_different_by_monthly_and_calls_and_less_payment_delay["cancelou"]
    .value_counts(normalize=True)
    .rename("After removing delay payments")
)

relation_cancellations_concat = pd.concat(
    [
        resume_cancellations,
        resume_cancellations_with_monthly_filter,
        resume_cancellations_with_monthly_filter_and_calls_filter,
        resume_cancellations_with_monthly_filter_and_calls_filter_and_payment_dalay_filter
    ],
    axis=1
).reset_index()

relation_cancellations_concat = relation_cancellations_concat.melt(
    id_vars="cancelou",
    var_name="Table",
    value_name="Cancelled"
)

relation_cancellations_concat["Status"] = relation_cancellations_concat["cancelou"].map({
    0.0: "Active Clients",
    1.0: "Cancelled Clients"
})
display(relation_cancellations_concat)

final_graph = px.bar(
    relation_cancellations_concat,
    x="Table",
    y="Cancelled",
    color="Status",
    barmode="stack",
    text_auto=".1%",
    title="Final graph",
    color_discrete_map={
        "Active Clients": "#3498DB",
        "Cancelled Clients": "#E74C3C"
    },
    labels={
       "Cancelled": "Cancelled (0/1)",
       "Table": "In accordance with the removal of bad things"
    }
)

final_graph.show()

```
